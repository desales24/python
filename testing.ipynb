```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

fig, ax = plt.subplots(figsize=(4, 6))

# Kotak-kotak diagram
boxes = [
    {"label": "Mulai", "xy": (0.5, 5), "color": "#B4B2A9"},
    {"label": "Proses Data", "xy": (0.5, 3), "color": "#85B7EB"},
    {"label": "Selesai", "xy": (0.5, 1), "color": "#5DCAA5"},
]

for box in boxes:
    rect = patches.FancyBboxPatch(
        (box["xy"][0] - 1, box["xy"][1] - 0.4),
        2, 0.8,
        boxstyle="round,pad=0.1",
        facecolor=box["color"],
        edgecolor="black"
    )
    ax.add_patch(rect)
    ax.text(box["xy"][0], box["xy"][1], box["label"],
             ha="center", va="center", fontsize=12)

# Panah penghubung
for i in range(len(boxes) - 1):
    y_start = boxes[i]["xy"][1] - 0.4
    y_end = boxes[i + 1]["xy"][1] + 0.4
    ax.annotate(
        "", xy=(0.5, y_end), xytext=(0.5, y_start),
        arrowprops=dict(arrowstyle="->", lw=1.5)
    )

ax.set_xlim(-1.5, 2.5)
ax.set_ylim(0, 6)
ax.axis("off")
plt.tight_layout()
plt.show()
```
